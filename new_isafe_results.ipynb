```python
import pandas as pd
import glob
import os
```

```python
# Directory where I-SAFE files are stored
input_dir = "/home/mrc.gm/hsaizonou/an_analysis/prep_isafe_inputs/isafe_results"  
```

```python
ISAFE_THRESHOLD = 0.25

results = []

for file in glob.glob(os.path.join(input_dir, "*.iSAFE.out")):

    fname = os.path.basename(file)
    base = fname.replace(".iSAFE.out", "")

    # Split and remove empty strings to handle double underscores
    parts = [p for p in base.split("_") if p]

    # Extract metadata
    population = parts[0]
    chrom = parts[1]
    start = parts[2]
    end = parts[3]
    region = f"{chrom}:{start}-{end}"

    # Load and clean data
    df = pd.read_csv(file, sep="\t", header=None, comment="#")
    df.columns = ["POS", "iSAFE", "DAF"]

    df = df[df["POS"] != "POS"]

    df["POS"] = df["POS"].astype(int)
    df["iSAFE"] = df["iSAFE"].astype(float)
    df["DAF"] = df["DAF"].astype(float)

    # Filter on iSAFE threshold
    df = df[df["iSAFE"] > ISAFE_THRESHOLD].copy()

    # Add metadata
    df["population"] = population
    df["region"] = region
    df["filename"] = fname

    results.append(df)

# Combine
combined_df = pd.concat(results, ignore_index=True)

# Rank globally
combined_df = combined_df.sort_values("iSAFE", ascending=False).reset_index(drop=True)
combined_df
```

```python
combined_df.to_csv("combined_isafe_results_af1_new_threshold_0.25.csv")
```
